```python

import pandas as pd
df_7 = pd.read_csv(r'C:\Nepal_Flood_Project\Data\final_flood_dataset.csv')


CITY_TO_DISTRICT_RENAME = {
    'kathmandu':  'Kathmandu',
    'pokhara':    'Kaski',
    'illam':      'Ilam',
    'manang':     'Manang',
    'biratnagar': 'Morang',
    'nepalgunj':  'Banke',
    'janakpur':   'Dhanusa',
}
df_7['location'] = df_7['location'].map(CITY_TO_DISTRICT_RENAME)


TERRAIN_MAP_7 = {
    'Kathmandu': 'Hilly', 'Kaski': 'Hilly', 'Ilam': 'Hilly',
    'Manang': 'Mountain', 'Morang': 'Terai', 'Banke': 'Terai', 'Dhanusa': 'Terai',
}
df_7['terrain'] = df_7['location'].map(TERRAIN_MAP_7)

print(df_7['location'].unique())
print(df_7.shape)
```

```python
df_70 = pd.read_csv(r'C:\Nepal_Flood_Project\Data\Districts_77\df_70_labeled_final.csv')
print('df_70 loaded:', df_70.shape)
```

```python

df_7_clean = df_7.drop(columns=['location_encoded', 'season_encoded'], errors='ignore')
df_70_clean = df_70.drop(columns=['GWETROOT', 'discharge_donor'], errors='ignore')


print('df_7 columns:', sorted(df_7_clean.columns.tolist()))
print('df_70 columns:', sorted(df_70_clean.columns.tolist()))
print('Same columns?', set(df_7_clean.columns) == set(df_70_clean.columns))
```

```python

print(df_70[['discharge', 'discharge_x', 'discharge_y']].head(10))
print('\nAre discharge_x and discharge_y identical?', (df_70['discharge_x'] == df_70['discharge_y']).all())
```

```python
mismatch = df_70[df_70['discharge_x'] != df_70['discharge_y']]
print(f'Mismatched rows: {len(mismatch)}')
print(mismatch[['location', 'date', 'discharge', 'discharge_x', 'discharge_y']].head(10))
```

```python
df_70 = df_70.drop(columns=['discharge_x', 'discharge_y'])
print('Cleaned. Columns now:', sorted(df_70.columns.tolist()))
```

```python
df_7_clean = df_7.drop(columns=['location_encoded', 'season_encoded'], errors='ignore')
df_70_clean = df_70.drop(columns=['GWETROOT', 'discharge_donor'], errors='ignore')

print('df_7 columns:', sorted(df_7_clean.columns.tolist()))
print('df_70 columns:', sorted(df_70_clean.columns.tolist()))
print('Same columns?', set(df_7_clean.columns) == set(df_70_clean.columns))
```

```python
# Combine into the final 77-district dataset
df_77 = pd.concat([df_7_clean, df_70_clean], ignore_index=True)
print(f'Combined dataset: {df_77.shape[0]} rows x {df_77.shape[1]} columns')
print(f'Unique districts: {df_77["location"].nunique()}')
print(df_77['location'].unique())
```

```python
from sklearn.preprocessing import LabelEncoder

le_location = LabelEncoder()
df_77['location_encoded'] = le_location.fit_transform(df_77['location'])

le_season = LabelEncoder()
df_77['season_encoded'] = le_season.fit_transform(df_77['season'])

le_terrain = LabelEncoder()
df_77['terrain_encoded'] = le_terrain.fit_transform(df_77['terrain'])

print('Districts encoded:', len(le_location.classes_))
print('\nFinal label distribution:')
print(df_77['flood_risk'].value_counts())
print(df_77['flood_risk'].value_counts(normalize=True).mul(100).round(1))
```

```python
df_77.to_csv(r'C:\Nepal_Flood_Project\Data\Districts_77\final_flood_dataset_77districts.csv', index=False)
print('Saved!')
```
