```python
import time
from geopy.geocoders import Nominatim

geolocator = Nominatim(user_agent="nepal_flood_project")
records = []

for d in DISTRICTS:
    try:
        loc = geolocator.geocode(f"{d} District, Nepal", timeout=10)
        if loc:
            records.append({'district': d, 'lat': loc.latitude, 'lon': loc.longitude})
            print(f'{d}: {loc.latitude}, {loc.longitude}')
        else:
            print(f'{d}: NOT FOUND — needs manual check')
        time.sleep(1)
    except Exception as e:
        print(f'{d}: FAILED - {e}')

districts_df = pd.DataFrame(records)
print(f'\nSuccessfully geocoded: {len(districts_df)} / {len(DISTRICTS)} districts')
```

```python
import requests

def get_max_elevation(lat, lon, offset=0.15):
    
    points = [
        (lat, lon),
        (lat + offset, lon), (lat - offset, lon),
        (lat, lon + offset), (lat, lon - offset),
    ]
    coords_str = "|".join([f"{la},{lo}" for la, lo in points])
    url = f"https://api.open-elevation.com/api/v1/lookup?locations={coords_str}"
    r = requests.get(url, timeout=15).json()
    elevs = [pt['elevation'] for pt in r['results']]
    return max(elevs)

def classify_terrain(max_elev):
    if max_elev < 500:
        return 'Terai'
    elif max_elev < 3000:
        return 'Hilly'
    else:
        return 'Mountain'

max_elevations = []
terrains = []
for _, row in districts_df.iterrows():
    try:
        max_elev = get_max_elevation(row['lat'], row['lon'])
        terr = classify_terrain(max_elev)
        max_elevations.append(max_elev)
        terrains.append(terr)
        print(f"{row['district']}: max {max_elev} m → {terr}")
    except Exception as e:
        max_elevations.append(None)
        terrains.append(None)
        print(f"{row['district']}: FAILED - {e}")
    time.sleep(1)

districts_df['max_elevation'] = max_elevations
districts_df['terrain'] = terrains
```

```python
STATION_COORDS = {
    'kathmandu':  (27.72, 85.32),
    'pokhara':    (28.21, 83.99),
    'illam':      (26.91, 87.93),
    'manang':     (28.67, 84.02),
    'biratnagar': (26.45, 87.27),
    'nepalgunj':  (28.05, 81.62),
}
STATION_TERRAIN = {
    'kathmandu': 'Hilly', 'pokhara': 'Hilly', 'illam': 'Hilly',
    'manang': 'Mountain', 'biratnagar': 'Terai', 'nepalgunj': 'Terai',
}

from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    dlat, dlon = radians(lat2 - lat1), radians(lon2 - lon1)
    a = sin(dlat/2)**2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon/2)**2
    return R * 2 * atan2(sqrt(a), sqrt(1 - a))

def nearest_station(row):
    same_terrain = [s for s, t in STATION_TERRAIN.items() if t == row['terrain']]
    candidates = same_terrain if same_terrain else list(STATION_COORDS.keys())
    dists = {s: haversine(row['lat'], row['lon'], *STATION_COORDS[s]) for s in candidates}
    return min(dists, key=dists.get)

districts_df['discharge_donor'] = districts_df.apply(nearest_station, axis=1)
print(districts_df[['district', 'terrain', 'discharge_donor']].to_string(index=False))
```

```python
import requests

def download_nasa_power(lat, lon, district_name, start='20000101', end='20251231'):
    url = "https://power.larc.nasa.gov/api/temporal/daily/point"
    params = {
        'parameters': 'T2M,PRECTOTCORR,RH2M,WS2M,GWETROOT',
        'community': 'AG',
        'longitude': lon, 'latitude': lat,
        'start': start, 'end': end, 'format': 'CSV'
    }
    r = requests.get(url, params=params, timeout=60)
    fname = fr'C:\Nepal_Flood_Project\Data\Districts_77\nasa_{district_name.replace(" ", "_")}.csv'
    with open(fname, 'w') as f:
        f.write(r.text)
    return fname

downloaded = []
failed = []
for _, row in districts_df.iterrows():
    try:
        fname = download_nasa_power(row['lat'], row['lon'], row['district'])
        downloaded.append(row['district'])
        print(f"Downloaded: {row['district']}")
    except Exception as e:
        failed.append(row['district'])
        print(f"FAILED: {row['district']} - {e}")
    time.sleep(1)

print(f'\nDownloaded: {len(downloaded)} / {len(districts_df)}')
if failed:
    print('Failed districts:', failed)
```

```python
# Quick sanity check on one file
import pandas as pd

def read_nasa_file(filepath, city_name):
    skip_rows = 0
    with open(filepath, 'r') as f:
        for i, line in enumerate(f):
            if '-END HEADER-' in line:
                skip_rows = i + 1
                break
    df = pd.read_csv(filepath, skiprows=skip_rows)
    df.columns = df.columns.str.strip()
    df['location'] = city_name
    return df

test_df = read_nasa_file(r'C:\Nepal_Flood_Project\Data\Districts_77\nasa_Jhapa.csv', 'Jhapa')
print(test_df.shape)
print(test_df.columns.tolist())
print(test_df.head())
```

```python
import os

all_districts_data = []
for _, row in districts_df.iterrows():
    district_name = row['district']
    filepath = fr'C:\Nepal_Flood_Project\Data\Districts_77\nasa_{district_name.replace(" ", "_")}.csv'
    if os.path.exists(filepath):
        df_d = read_nasa_file(filepath, district_name)
        all_districts_data.append(df_d)
    else:
        print(f'Missing file: {district_name}')

df_70 = pd.concat(all_districts_data, ignore_index=True)
print(f'Combined 70-district dataset: {df_70.shape[0]} rows x {df_70.shape[1]} columns')
```

```python
import numpy as np
df_70['date'] = pd.to_datetime(
    df_70['YEAR'].astype(str) + '-' + df_70['DOY'].astype(str),
    format='%Y-%j'
)
df_70['month'] = df_70['date'].dt.month
df_70['year']  = df_70['date'].dt.year
df_70['day']   = df_70['date'].dt.day

def get_season(month):
    if month in [3, 4, 5]:
        return 'pre_monsoon'
    elif month in [6, 7, 8, 9]:
        return 'monsoon'
    elif month in [10, 11]:
        return 'post_monsoon'
    else:
        return 'winter'

df_70['season'] = df_70['month'].apply(get_season)
df_70 = df_70.rename(columns={
    'T2M': 'temperature', 'PRECTOTCORR': 'rainfall',
    'RH2M': 'humidity', 'WS2M': 'wind_speed'
})
df_70 = df_70.drop(columns=['YEAR', 'DOY'], errors='ignore')
df_70 = df_70.replace(-999, np.nan)
df_70 = df_70.sort_values(['location', 'date']).reset_index(drop=True)

print(df_70[['location','date','temperature','rainfall','humidity','wind_speed','GWETROOT']].head())
```

```python
import re
import os

DHM_FOLDER = r"C:\Nepal_Flood_Project\Data\DHM Data"

STATION_CITY = {
    '550.05': 'kathmandu',
    '695':    'biratnagar',
    '280':    'nepalgunj',
    '430.5':  'pokhara',
    '795':    'illam',
    '439.7':  'manang',
}

def read_dhm_file(filepath):
    records = []
    with open(filepath, 'r', errors='ignore') as f:
        for line in f:
            line = line.strip()
            m = re.match(r'(\d{2}/\w+/\d{4}),(\S+)', line)
            if m:
                try:
                    date = pd.to_datetime(m.group(1), format='%d/%b/%Y').date()
                    discharge = float(m.group(2))
                    records.append({'date': date, 'discharge': discharge})
                except:
                    pass
    return pd.DataFrame(records)

discharge_lookup = {}
for station_no, city in STATION_CITY.items():
    filepath = os.path.join(DHM_FOLDER, f'DFL_{station_no}.txt')
    if os.path.exists(filepath):
        df_st = read_dhm_file(filepath)
        discharge_lookup[city] = df_st.set_index('date')['discharge'].to_dict()
        print(f'DHM loaded {city}: {len(df_st)} records')
```

```python
def get_discharge_district(row):
    donor = row['discharge_donor']
    row_date = pd.Timestamp(row['date']).date()
    donor_vals = discharge_lookup.get(donor, {})
    if not donor_vals:
        return 0
    return donor_vals.get(row_date, np.mean(list(donor_vals.values())))

df_70['discharge'] = df_70.apply(get_discharge_district, axis=1)
print('Discharge feature added!')
print(df_70.groupby('location')['discharge'].mean().round(1).head(15))
```

```python
def assign_flood_risk_70(row):
    row_date = pd.Timestamp(row['date']).date()
    district = row['location']
    terrain = row['terrain']

    if terrain == 'Mountain':
        return 'Low'

    if row_date in bipad_high_dates.get(district, set()):
        return 'High' if terrain == 'Terai' else 'Medium'
    if row_date in bipad_medium_dates.get(district, set()):
        return 'Medium'
    return 'Low'

df_70['flood_risk'] = df_70.apply(assign_flood_risk_70, axis=1)
risk_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
df_70['flood_risk_label'] = df_70['flood_risk'].map(risk_mapping)

print(df_70['flood_risk'].value_counts())
print(df_70['flood_risk'].value_counts(normalize=True).mul(100).round(1).astype(str) + '%')
```
